Clones the published repository and prints the exact revision. No Census key is stored in this notebook. Cached inputs support runs without a key; fresh MOEs require CENSUS_API_KEY.

```python
# Audited BluePrint — run this cell only; older notebook cells are historical.
from pathlib import Path
import subprocess, sys
project = Path('/content/blueprint_audited_repository_20261018')
if not (project/'.git').exists():
    subprocess.run(['git','clone','https://github.com/moulikpagarani/market-intelligence-nonprofit.git',str(project)],check=True)
print('Repository revision:', subprocess.check_output(['git','rev-parse','HEAD'],cwd=project,text=True).strip())
subprocess.run([sys.executable,'-m','pip','install','-q','-r',str(project/'requirements.txt')],check=True)
subprocess.run([sys.executable,str(project/'scripts/blueprint_pipeline.py')],check=True)
subprocess.run([sys.executable,'-m','unittest','discover','-s',str(project/'tests'),'-v'],check=True)
print('AUDITED COLAB RUN PASSED; MOE limitations remain explicit in the report.')
print((project/'reports/week7_technical_documentation.md').read_text())
```
